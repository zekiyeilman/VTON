## 1. Google Drive'ı Bağla ve Kodu Hazırla

```python
import os, subprocess
from google.colab import drive
drive.mount("/content/drive")

CALISMA = "/content/IDM-VTON"
DRIVE_WEIGHTS = "/content/drive/MyDrive/IDM-VTON-weights"

# Model ağırlıklarının Drive'da olduğunu doğrula
if os.path.exists(DRIVE_WEIGHTS):
    print("✅ MÜKEMMEL: Model ağırlıkları Google Drive'ınızda hazır bekliyor! (İndirme YOK)")
else:
    print("⚠️ Uyarı: /content/drive/MyDrive/IDM-VTON-weights bulunamadı, Drive'ınızı kontrol edin.")

# Kod ve yardımcı ckpt dosyalarını hızlıca hazırla (~10 saniye)
if not os.path.exists(CALISMA):
    from huggingface_hub import snapshot_download
    print("Hafif arayüz kodları alınıyor...")
    snapshot_download("yisol/IDM-VTON", repo_type="space", local_dir=CALISMA)
print("✅ Kod ve modeller hazır.")
```

## 3. Modeli Başlat ve Public Link Üret (Canlı Bağlantı)

Bu hücre Drive'daki hazır ağırlıkları okur ve birkaç dakika içinde public `https://xxxx.gradio.live` linkinizi üretir.

```python
import os, re

CALISMA = "/content/IDM-VTON"
BASE_PATH = "/content/drive/MyDrive/IDM-VTON-weights"

# app.py dosyasını Drive yoluna ve public linke ayarla
app_path = os.path.join(CALISMA, "app.py")
src = open(app_path, encoding="utf-8").read()
src = re.sub(r"base_path\s*=\s*['\"][^'\"]+['\"]", f"base_path = '{BASE_PATH}'", src, count=1)
src = re.sub(r"image_blocks\.launch\([^)]*\)", "image_blocks.launch(share=True, debug=True, show_error=True)", src)
open(app_path, "w", encoding="utf-8").write(src)

print("🚀 Model Drive üzerinden yükleniyor...")
%cd /content/IDM-VTON
!python app.py
```
